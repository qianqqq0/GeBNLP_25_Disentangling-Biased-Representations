```python
import pandas as pd

data = pd.read_csv("dataset/test_senti_data_6.csv")
```

```python
race_terms_mapping = {
    "African": "European",
    "European": "African",
    "Asian": "European",
    "Hispanic": "European",
    "Black": "White",
    "White": "Black"
}
```

```python
def has_race_word(text):
    words = text.split()
    for word in words:
        if word in race_terms_mapping:
            return True
    return False

def generate_race_counterfactual(text):
    if has_race_word(text):
        words = text.split()
        counterfactual_words = []
        for word in words:
            if word in race_terms_mapping:
                counterfactual_words.append(race_terms_mapping[word])
            else:
                counterfactual_words.append(word)
        return " ".join(counterfactual_words)
    return None

```

```python
data['race_counterfactual_text'] = data['text'].apply(generate_race_counterfactual)
```

```python
data_new = data[data['race_counterfactual_text'].notna()]
```

```python
data_new.to_csv('counterfactual_race.csv', index=False)
```
