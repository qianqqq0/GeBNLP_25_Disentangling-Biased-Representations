```python
import pandas as pd

data = pd.read_csv("dataset/test_senti_data_6.csv")
```

```python
term_mapping = {
    "man": "woman", "men": "women", "boy": "girl", "boys": "girls",
    "father": "mother", "fathers": "mothers", "son": "daughter", "sons": "daughters",
    "brother": "sister", "brothers": "sisters", "husband": "wife", "husbands": "wives",
    "uncle": "aunt", "uncles": "aunts", "nephew": "niece", "nephews": "nieces",
    "emperor": "empress", "emperors": "empresses", "king": "queen", "kings": "queens",
    "prince": "princess", "princes": "princesses", "duke": "duchess", "dukes": "duchesses",
    "lord": "lady", "lords": "ladies", "knight": "dame", "knights": "dames",
    "waiter": "waitress", "waiters": "waitresses", "actor": "actress", "actors": "actresses",
    "god": "goddess", "gods": "goddesses", "policeman": "policewoman", "policemen": "policewomen",
    "postman": "postwoman", "postmen": "postwomen", "hero": "heroine", "heroes": "heroines",
    "wizard": "witch", "wizards": "witches", "steward": "stewardess", "stewards": "stewardesses",
    "guy": "gal", "guys": "gals", "male": "female", "males": "females",
    "dad": "mom", "dads": "moms", "he": "she", "his": "her",
    "him": "her", "himself": "herself"
}

reverse_mapping = {v: k for k, v in term_mapping.items()}
term_mapping.update(reverse_mapping)
```

```python
def has_gender_word(text):
    words = text.lower().split()
    for word in words:
        if word in term_mapping:
            return True
    return False

def generate_counterfactual(text):
    if has_gender_word(text):
        words = text.lower().split()
        counterfactual_words = []
        for word in words:
            if word in term_mapping:
                counterfactual_words.append(term_mapping[word])
            else:
                counterfactual_words.append(word)
        return " ".join(counterfactual_words)
    return None

```

```python
data['counterfactual_text'] = data['text'].apply(generate_counterfactual)
```

```python
data_new = data[data['counterfactual_text'].notna()]
```
