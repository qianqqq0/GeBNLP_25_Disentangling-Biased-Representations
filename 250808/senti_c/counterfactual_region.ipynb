```python
import pandas as pd

data = pd.read_csv("dataset/test_senti_data_6.csv")
```

```python
region_terms_mapping = {
    "Paris": "Beijing",
    "New York": "Beijing",
    "Tokyo": "London",
    "London": "Tokyo",
    "Beijing": "New York",
    "Sydney": "Beijing",
    "Mumbai": "New York",
    "Berlin": "Mumbai",
    "Vienna": "Shanghai"
}
```

```python
def has_region_word(text):
    words = text.split()
    for word in words:
        if word in region_terms_mapping:
            return True
    return False

def generate_region_counterfactual(text):
    if has_region_word(text):
        words = text.split()
        counterfactual_words = []
        for word in words:
            if word in region_terms_mapping:
                counterfactual_words.append(region_terms_mapping[word])
            else:
                counterfactual_words.append(word)
        return " ".join(counterfactual_words)
    return None

```

```python
data['region_counterfactual_text'] = data['text'].apply(generate_region_counterfactual)
```

```python
data_new = data[data['region_counterfactual_text'].notna()]
```

```python
data_new.to_csv('counterfactual_region.csv', index=False)
```
