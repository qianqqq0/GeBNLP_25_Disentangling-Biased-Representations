```python
import pandas as pd

data = pd.read_csv("dataset/test_senti_data_6.csv")
```

```python
religion_terms_mapping = {
    "Christian": "Muslim",
    "Buddhist": "Christian",
    "Muslim": "Christian",
    "Jewish": "Christian",
}
```

```python
def has_religion_word(text):
    words = text.split()
    for word in words:
        if word in religion_terms_mapping:
            return True
    return False

def generate_religion_counterfactual(text):
    if has_religion_word(text):
        words = text.split()
        counterfactual_words = []
        for word in words:
            if word in religion_terms_mapping:
                counterfactual_words.append(religion_terms_mapping[word])
            else:
                counterfactual_words.append(word)
        return " ".join(counterfactual_words)
    return None
```

```python
data['religion_counterfactual_text'] = data['text'].apply(generate_religion_counterfactual)
```

```python
data_new = data[data['religion_counterfactual_text'].notna()]
```

```python
data_new.to_csv('counterfactual_religion.csv', index=False)
```
